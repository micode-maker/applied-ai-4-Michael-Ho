## Reflection

My agent, the "Forgiving Giant", finished 1st out of 23 in the round-robin tournament with 12,439 points but got knocked out in round 3 of the bracket. Its win rate was only 55.3%, which shows that scoring the most points and winning the most matches are not the same thing. The biggest takeaway for me was that how you measure success changes which strategy looks best. 

### Tournament Performance Analysis

My agent’s final rank was No. 1 out of 23, with 12,439 points, 1,378 points ahead of #2. It performed alright in the bracket; it beat Ruthless Optimizer 162-92, tied against My Evolved Agent 300-300, and won the sudden death. However, it lost to Chaos Agent 285-310 in round 3 and got knocked out. In the round-robin, it did really well and better than I expected; my agent ranked 1st with the highest average score of 188.9 but only had a win rate of 55.3%. The results between bracket and round-robin were quite different. In the bracket, winning a match is based on outscoring the opponent in front of you. My agent did well in the first two rounds but lost a close, high-scoring match against Chaos Agent in round 3. round-robin adds up each agent's points from 132 matches and bases your rank on that. My agent is able to earn a lot of points each match but has a tight win-loss margin.

### Strategy Analysis

My "Forgiving Giant" agent placed 1st; it invests 92% of the time in the first round, looks back over the opponent’s last 5 moves, and if they invest less than around 50% of the time, it always undercuts. ‘My Evolved Agent’ came 2nd; it cooperates for the first 3 rounds. After that, it always cooperates if the opponent has cooperated over 80% of the time and never forgives an opponent who cooperated less than 50% of the time. The ‘Stable Cooperator’ agent placed 3rd. It has an “Always Cooperate” type of strategy that forgives around 20% of the time. Although it has a win rate of 5.3%, it still came in 3rd. Comparing the two agents that came 2nd and 3rd with my agent, my agent opens with cooperation 92% of the time and always reciprocates. It has a 76% forgive gene and a 5-round memory rule alongside an “undercut if opponent cooperates less than around 50% of the time” rule. However, the undercut rule is checked first, so the forgive rule rarely gets used against opponents that undercut. All three strategies cooperated by default; this repeated cooperation earns each strategy around 3 points a round, providing way more points than one-off undercuts do.

### Bracket vs Round-Robin

The bracket champion was not the same as the round-robin champion. The bracket champion was the ‘Adaptive Survivor’ who came 10th in the round-robin tournament. The round-robin champion was my ‘Forgiving Giant’ who got knocked out in round 3 of the bracket tournament. This teaches me that a bracket format tournament has high variance. There were three ties that were decided by sudden death; one bad draw ends your run. Total points matter less in a bracket when compared to round-robin tournaments. I believe the round-robin format reflects true quality because it measures which agent produces the most total value rather than just beating an opponent.

### Superlative Awards

The award that my agent won was ‘Most Cooperative’ with an average score of 188.9. The trade-off is that my agent was on the losing end of the Biggest Upset and Glass Cannon. Although my agent earned a lot of points being the most cooperative, being forgiving can be exploited. Personally, I would rather have a high total score than a high win rate. A high total score means that the agent brings value, whereas a high win rate can mean that the agent wins even at the cost of bringing in less value.

### Generalization Lesson

After being trained on 10 opponents in the last unit and then facing 20+ opponents here, forgiveness carried over well as most of the field was willing to cooperate. If I could redesign my agent for this environment in particular, I would fix the responsiveness check to read my moves instead of the opponent’s moves and possibly score fitness on both the total score and win rate.

### Game Theory Insights

I believe that there is no single best strategy, because how well a strategy does depends on who else is in the field. In a field of cooperators, a cooperative strategy like mine scores the most, but in a field of aggressive agents, it becomes a price war. Real-world situations that mirror this tournament include supplier relationships, price wars, and repeated negotiations. For example, a business that works with the same supplier does better by building trust rather than by squeezing the most out of every deal, just like how my agent did better by cooperating over 100 rounds rather than undercutting.
